Side note: we can make the output meaningful for a human

````python
class Segment:
    #definition
    def __init__(self, x1, y1, x2, y2):
        self.ptA = (x1,y1)
        self.ptB = (x2,y2)
        
        return
    
    def length(self):
        return ((self.ptA[0] - self.ptB[0])**2. + (self.ptA[1] - self.ptB[1])**2.)**.5
    
    def __str__(self):
        return f"{self.ptA}--{self.ptB}"
    
    def __lt__(self, s):
        return self.length() < s.length()
    
    def __eq__(self, s):
        return self.length() == s.length()
    
    def __repr__(self):
        return self.__str__()
````

````python
rand_segs = [ Segment(np.random.random(),np.random.random(),
                      np.random.random(),np.random.random()) for i in range(10)]
````

### Naming (and other) conventions
- Class names should be in UpperCaseCamelCase
- Member attributes (variables) should be all lower case (and meaningful)
- Single leading underscore (`_variable`) denotes a class member attribute is meant for internal use only (private = accessed and modified only by the class member functions, usually to keep consistency). 
- Single trailing underscore naming convention is used to avoid conflicts with Python keywords. When the most fitting name for a variable is already taken by a keyword, appending a single underscore convention is followed to break the naming conflict. Typical example includes using `class_` or other keywords as variables.
- Double leading and trailing underscore (`__var__`) names are _reserved_ for special use like the `__init__` method for object constructors, or `__str__` method to represent the object as string and operator overloading. 
- Double leading underscore (`__var`). The use of double underscore (`__`) in front of a name (specifically a method name) is not a convention; it has a specific meaning to the interpreter. Python mangles these names and it is used to avoid name clashes with names defined by subclasses.
