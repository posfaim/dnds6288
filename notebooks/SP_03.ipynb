### Sorting

Sorting lists is possible using the `sort()` method of lists or the built-in `sorted()` function:





````python
l = [34,1,2,78,3]

sl = sorted(l) # creates a new list
print("Sorted list:",sl, "original:",l)
print()

l.sort() #works in place
print(l)
````

The `key` argument allows us to sort based on any property.

We need to define an `f` function that takes an element of the list as input, and returns the property based on we would like to sort the list. Then we use `f` as the `key` for sorting:

````python
def f(x):
    return x%7

L = [15, 3, 11, 7] 
print("Sorted with key:", sorted(L, key = f ))
````

Lambda functions provide a more concise way to do the same thing:

````python
L = [15, 3, 11, 7] 
print("Sorted with key:", sorted(L, key = lambda x: x%7) )
````

### Dimension and shape

Technically, an array is stored in the computer's memory as a one dimensional sequence of data. For example, it can be a sequence of ints:
<code>
            |--int--|--int--|--int--|--int--|--int--|--int--|--int--|--int--|
</code>

However, this one dimensional sequence can be indexed in different ways. The simplest indexing provides a one dimensional array `a`:
<code>
    index   |0      |1      |2      |3      |4      |5      |6      |7      |       
            |--int--|--int--|--int--|--int--|--int--|--int--|--int--|--int--|
</code>
To access the leftmost element: `a[0]`.

But we can view the data as a multi-dimensional array using two or more indices, for example:
<code>
row index   |0                              |1                              |
col index   |0      |1      |2      |3      |0      |1      |2      |3      |
            |--int--|--int--|--int--|--int--|--int--|--int--|--int--|--int--|
</code>
This indexing creates a two dimensional array `b`, with 2 rows and 4 columns. To access the leftmost element: `b[0][0]`. And `a[6]` is the same as `b[1][2]`.

The dimension of an array is the number of indices used to identify an element, the shape of an array is a tuple containing the size of the array along each axis:

````python
a = np.array([1,2,3,4,5,6,7,8])     #one dimensional array
b = np.array([[1,2,3,4],[5,6,7,8]]) #two dimensional array
print(np.ndim(a),np.ndim(b))
print(a.shape, b.shape)
````

````python
a[np.argsort(-a)]
````

You can also reshape an array:

````python
c = a.reshape(2,2,2)
print(c)
print(np.ndim(c), c.shape)
````

This is a very fast operation, as it does not require to copy the data, just to set a new indexing scheme!

`a` and `c` are not the same array, but they share the same data. In NumPy parlance: `c` and `a` are two views of the same data.



````python
print(a is c) #not the same object
print(a)
c[0][0][0] += 100 #but the share data => changing c changes a too
#or using alternative indexing style
c[0,0,0] += 100
print(a)
````

Since you are reindexing the same amount of data, the shape of arrays can change, but the total number of items cannot:

````python
print(a.reshape(4, 2))  # 4*2   == 8
print(a.reshape(4,1,2)) # 4*1*2 == 8
print(a.reshape(1, 8))  # 1*8   == 8
````

### Arrays are mutable

Arrays are of course mutable, i.e., you can change the values contained in the array without NumPy making a completely new copy of the array.

````python
a = np.ones(5) # creates an array of ones
b = a    # b and a point at the same object
b[0] = 9 #changing b changes a, just like in the case of lists
print("a=", a)
print()

c = np.copy(a) #makes a new copy of the array in the memory 
c[0] = 0 #changing c does not change a
print("a=",a,"\nc=",c)
````

Most array operations and mathematical functions implemented in NumPy are applied to arrays elementwise:


### Loops vs array operations (important!)

Working with arrays you should vectorize your algorithm, meaning that you avoid explicit `for` loops. Array operations off-load Python loops to compiled C code, leading to large performance improvements.

Consider:

````python
import timeit
import math

x = np.arange(1000000)

def f(x):
    y = np.zeros(x.shape)
    for i in range(len(x)):
        y[i] = math.cos(x[i]) + x[i] ** 2
    return y

#using a for loop
print(f(x))
#same thing as an array expression
print(np.cos(x) + x ** 2)  

print(timeit.timeit(lambda: f(x), number=1))
print(timeit.timeit(lambda: np.cos(x) + x ** 2, number=1))
````

## Side note: measuring runtime

In the previous example we measured the runtime of our code using the `timeit` module. This module provides useful functionality to test and analyze your code. For example, if you are planning to do some computationally intensive task on a large data set, you might want to measure how fast your code is on a smaller data sample and estimate how long it will take for the large data set.

The simplest way to use `timeit` is to use the eponymous function of the module: `timeit.timeit`. It can measure the runtime of a function that does not take any arguments:

````python
def func():
    s = 0
    for i in range(1000):
        s += i**2
    return s

n = 1000
T = timeit.timeit(func, number=n)
print("It takes %f seconds to run func %d times."%(T,n))
````

The parameter `number` tells `timeit` the number of times it should execute the `func` function.

What if we would like to measure a function that takes arguments? We can work around this using lambda functions!

````python
def func(N):
    s = 0
    for i in range(N):
        s += i**2
    return s

n = 1000
T = timeit.timeit(lambda : func(1000), number=n)
print("It takes %f seconds to run func %d times."%(T,n))
````

With `lambda: func(1000)` we define a lambda function that does not take any arguments.

Try running the above cell a few times! You can see that the exact runtime changes each time you run it. Isn't our code deterministic, shouldn't it always take the same time? There are two sources of uncertainty:
* The accuracy of the clock that we use to measure the runtime, which is typically around 1 ms. If your code runs around the order of 1 ms or faster, the measurement will be inaccurate, the same way you cannot accurately a grain of rice on a kitchen scale. You should set `number` so that the total runtime is on the order of seconds.
* During the time you are running your code, your computer and python are doing a bunch of things in the background. The operating system has to schedule these tasks and this will slow down your code.

To see how this second source of uncertainty affects runtime, we can use the `timeit.repeat` function, which simply repeatedly calles `timeit.timeit` and returns the runtimes in a list:

````python
runtimes = timeit.repeat(lambda : func(1000), number=1000, repeat=5)
print("runtimes:",runtimes)
print("The best case runtime is no more than",min(runtimes))
````

The minimum runtime is an estimate of how fast your code could be without interference by other processes.

In a jupyter notebook you can also measure the runtime of code cell using the magic command:

````python
%%timeit -r 5 -n 1000
func(1000)
````

This prints out the average runtime of a single execution and the standard deviation. Single exection here means running `func` once, i.e., deviding the output of `timeit.timeit` by `number`.

## Broadcasting

[Video.](https://ceu.cloud.panopto.eu/Panopto/Pages/Viewer.aspx?id=9c6c3467-f591-4112-901d-adb800d4d4b9)

For elementwise operations between arrays `a` and `b`, the shape of `a` and `b` should be the same.

What happens with array operations if two arrays have different shape?

````python
a = np.array([1.,2.,3.])
b = np.ones((3,3))
print("a=",a,a.shape,"\nb=", b, b.shape)
print("2 * a =\n", 2 * a)
print("a + 1 =\n", a + 1)
print("a + b =\n", a + b)
````

How does numpy deal with different shapes? This is called broadcasting.

A very simple case is when we multiply the array `a` with the number `2`. You can think of `2` as an array of shape `(1,)`. What NumPy does is that it multiplies all elements with `2`, you can think of this as `2` being stretched out to the same shape and then the two arrays are multiplied elementwise:
<code>
                          2*[1 2 3]  --> [2 2 2] * [1 2 3] = [2 4 6]
</code>

When similarly the one dimensional array `a` is stretched out in the missing dimension to match the shape of `b` before calculating `a+b`
<code>
                           [[1 1 1]       [[1 2 3]    [[1 1 1]    [[2 3 4]
                 [1 2 3] +  [1 1 1]   -->  [1 2 3]  +  [1 1 1]  =  [2 3 4]
                            [1 1 1]]       [1 2 3]]    [1 1 1]]    [2 3 4]]
</code>

General rule:

In order to broadcast, the size of the axes for both arrays in an operation must either be the same size or one of them must be one.

Along the axes of size one (or in missing dimensions) the array is stretched in order match the size of the other array.

## (Pseudo-)Random numbers

[Video.](https://ceu.cloud.panopto.eu/Panopto/Pages/Viewer.aspx?id=c86f963a-2fa3-4f25-b739-acb600ef5438)

Our computers are deterministic machines; therefore they cannot generate true random numbers. Instead they can produce deterministic sequences of numbers that in many ways behave very closely to random numbers. These are called pseudo-random numbers.

Random seed: Through the seed we can set an initial value of the pseudo-random sequence, since it is deterministic, the same seed produces the same sequence. Let's try this out with `np.random.random` which generates a random float between 0 and 1.

````python
np.random.seed(42)
a = np.random.random()
print(a)
````

No matter how many times you run the previous cell, it will produce the same "random" number, since we always re-seed the sequence. But if we run `np.random.random` without re-seeding, we always get a different number:

````python
a = np.random.random()
print(a)
````

It is possible to get arrays filled with random numbers:

````python
a = np.random.random(10)
print(a)
b = np.random.random((3,3))
print(b)
````

As a small example, let's simulate a biased coin flip:

````python
N = 10  # number of times we flip the coin
p = 0.2 # the coin is biased, we get heads with probability p
a = np.random.random(N) # generate N uniform random numbers
for x in a:
    if x < p: 
        print("Heads")
    else:
        print("Tails")
````

Or instead of a for loop, we can use the `<` operator on an array elementwise (also, rules of broadcasting apply):

````python
print(a<p)
````

where `True` is heads.

We are not restricted to the uniform distribution:

````python
print(np.random.randint(0, 10, size=10))
print(np.random.normal(1, 2, size=(2,2))) # mean =1, standard deviation= 2
````

There are many other distributions, check out the documentation.

A common task is to randomly choose from a list of posibilities:

````python
print(np.random.choice(5, 10)) # randomly choose integer from 0 to 4
print(np.random.choice(5, 3, replace=False)) # same thing without replacement
print(np.random.choice(5, 10, p=[0.1, 0, 0.3, 0.6, 0])) # same thing with non-uniform probability
c = ["Budapest", "Pécs", "Debrecen", "Miskolc", "Szeged"]
print(np.random.choice(c,1)) # randomly choose element from list
````

## Masking

We can mask out elements by indexing via boolean arrays. To mask out elements from an array `a`, use create a bool array `mask` with the same shape. If `mask` is used as an index only elements where `mask` is `True` are returned. 

````python
a = np.random.random(6)
mask = np.zeros(6,dtype=bool) # all values are False
mask[0:2] = True
print("a =", a)
print("mask =", mask)
print("a[mask] =", a[mask])
````

## Histogram

We can simply calculate a histogram of the elements of an array.

For example, let's generate 10,000 random numbers and get their histogram using 3 evenly spaced bins:

````python
a = np.random.random(10000)
counts,bin_edges = np.histogram(a, bins=3)
print(a)
print(counts)
print(bin_edges)
````

The `np.histogram` returns two values: the counts in each bin and the location of the edges of the bins. If `bins` is specified by an integer, the range of the bins is determined by the minimum and maximum value of `a`, and the bins are evenly distributed.

We can also directly specify the edges of the bins:

````python
a = np.random.random(10000)
counts, bin_edges = np.histogram(a, bins=[0, 0.75, 1.0])
print(counts)
````

## Histograms

[Video.](https://ceu.cloud.panopto.eu/Panopto/Pages/Viewer.aspx?id=f06c802f-7cec-4464-b9a3-acbb00f39971)

The `hist(data)` function can both calculate and plot a histogram from a sample `data`.

It has a `bins` attribute that works just like NumPy's `histogram` function: you can specify the number of evenly spaced bins or the edges of the bins.

If the `density` attribute is set to `True`, the counts in each bin are divided by the width of the bin, this way you get the probability density function. Let's try this out!

````python
fig, axes = plt.subplots(nrows=1, ncols=2, figsize=(12,4))

#generate a large sample of normally distributed numbers with mean 2 and standard deviation 1
x = np.random.normal(2,1,10000)

#we will compare the histogram to the PDF of the theoretical normal distribution
xs = np.linspace(-2,5.,50)
mean     = 2
variance = 1
normal_pdf = 1./(2.*np.pi*variance)**.5*np.exp(-((xs-mean)**2.)/2./variance)

true_or_false = True
for ax in axes:
    ax.hist(x, 50, density=true_or_false, facecolor='g',alpha=1.) #try changing facecolor and alpha
    ax.plot(xs,normal_pdf,'--',linewidth=5)
    ax.set_xlabel('x')
    ax.set_ylabel('y')
    ax.set_title('Density = '+str(true_or_false))
    true_or_false = not true_or_false

````

Now let's look at a two dimensional normal distribution:

````python
n_points = 100000

x = np.random.normal(0, 1, size=n_points)
y = np.random.normal(0, 2, size=n_points)

plt.plot(x,y,'o');
````

## 03 Dealing with dates and times

[Video.](https://ceu.cloud.panopto.eu/Panopto/Pages/Viewer.aspx?id=ed8331ab-b29f-40ee-99c0-acbb010f836f)

Timestamps can be surprisingly tricky to deal with in programs. For example:

* How many days are between October 26, 1985 and today?

To answer this properly, you need full information on the calendar, including leap years, time zones, etc.

### Datetime

Fortunately, Python has builtin support for date and time data using the [datetime](http://docs.python.org/2/library/datetime.html) module. It's not trivial to use but it works.

Let's answer the above question. `datetime` gives us useful date objects, `date`, `time`, and `datetime`. These let us store a date, a time of day, or both, respectively.

````python
import datetime

D1 = datetime.date(1985, 10, 26)
T1 = datetime.time(12, 0, 0) # noon
DT = datetime.datetime(1985, 10, 26, 12, 0, 0)

# Typically you want to work with datetime because you can
# omit the time values and then it defaults to midnight.
D = datetime.datetime(1985,10,26)
````

Once you have a `datetime` object you can do fancy things with it:

````python
print("The year was %i and the day was %i." % (D.year, D.day))

print ("The day of the week was %i." % (D.weekday()))
print ("(Monday = 0, ..., Sunday = 6.)")
````

You can get the current date and time with

````python
Dnow = datetime.datetime.now()
print(Dnow)
````

We can format the time using for example strftime (all information about the format [ here](https://docs.python.org/2/library/datetime.html#strftime-strptime-behavior)):

````python
Dnow.strftime("%I:%M%p") 
````

More importantly, these objects support math operations that are meaningful for time. For example, we can mention our original question:

````python
print (Dnow - D)
````

What this does is create another data object, called a `timedelta` object:

````python
dt = Dnow - D
print(type(dt))
print("There are %i days between then and now." % dt.days)
````

`timedelta` encodes time intervals. This allow us to do more operations:

````python
interval = datetime.timedelta(days=100,hours=12) # 100.5 days

soon = datetime.datetime.now() + interval # addition!

print ("In %d days and %d seconds it will be %s." % (interval.days, interval.seconds, soon))
````

OK, the real work comes when reading and writing **timestamps**. We need to be able to understand how `2012-04-26` means the same date as `April 26, 2012` does.

`datetime` provides us tools to read and write such timestamps. Let's first define two different timestamps

````python
ts1 = "2012-04-26"
ts2 = "January 5, 1978"
````

We can now use a function to parse a string for a time given a string representing a time format. This uses a function called `strptime()` (read it as "**str**ing **p**arse" **time**"). 

Here we go.

````python
d1 = datetime.datetime.strptime(ts1, "%Y-%m-%d" )
print(d1)
# now we can work with the datetime object, e.g.:
print(d1 + datetime.timedelta(days=-7))
````

The string `"%Y-%m-%d"` encodes the timestamp format we were looking for. A four-digit year (`%Y`), a dash (`-`), a two-digit month number (`%m`), another dash, and then a day number (`%d`).

Now ts2 incorporates the name of a month, so that format string is a little different (`%B` means the full month name).

````python
d2 = datetime.datetime.strptime( ts2, "%B %d, %Y" )
print(d2)
print(d2 - datetime.timedelta(days=-7))
````

There's a huge number of ways to build a format string. Best is to look up the [documentation](http://docs.python.org/2/library/datetime.html#strftime-strptime-behavior).

In addition to to `strptime` is another function, `strftime` (string format time) that does the opposite: it takes a `date` or `datetime` and returns a timestamp format string.

````python
s_before = "Jan 19, '89"
d = datetime.datetime.strptime("Jan 19, '89", "%b %d, '%y")
s_after  = d.strftime("%Y-%m-%d")
print (s_before, "--->", s_after)
````

Datetime is extremely useful, because different data sources encode times in different ways. Some formats are easy for humans to read, but I like the standard `%Y-%m-%d %H:%M:%S` UNIX-style timestamp because it _sorts nicely_.

## Side note: "The epoch".

[Video.](https://ceu.cloud.panopto.eu/Panopto/Pages/Viewer.aspx?id=79cc9d15-5217-4e73-a4c7-acbb0112da8f)

Sometimes you see a date that looks weird:

````python
import time # another time module!

print(time.time())
````

This function is another way to get the _current time_ but it's encoded using a single number: **the number of seconds since the "epoch"**. Let's explore:

````python
t = time.time()
y = t / 60 / 60 / 24 / 365.25 # oops, leap years!
print(y)
````

OK, what the heck happened back then?

````python
days_delta = time.time() / 60 / 60 / 24
print (datetime.datetime.now() - datetime.timedelta(days=days_delta))
````

The epoch ("epoch" means "reference date") is the [UNIX epoch](http://en.wikipedia.org/wiki/Unix_time), Jan 1, 1970.

* `time.time()` returns the number of seconds since 00:00:00 [Coordinated Universal Time](https://en.wikipedia.org/wiki/Coordinated_Universal_Time) (UTC), Thursday, 1 January 1970, not counting leap seconds.

Why 1970? Because of these guys:

These numeric timestamp formats are very useful when it's too expensive to use a complex library like `datetime`.
