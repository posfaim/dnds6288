### 01 Reshaping arrays

Take the one dimensional array `a` and create a list that contains all possible two dimensional arrays that can be formed by reshaping `a`.

<details><summary><u>Hint.</u></summary>
<p>

The reshaped array has to have the same number of elements as the original. For example, you can reshape a one dimensional array `a` of shape `(6,)` to a two dimensional array of `(3,2)` using `a.reshape((3,2))` because `6==3*2`.

When reshaping an array you can set one of the new dimensions to `-1` in which case numpy automatically selects the correct value if possible. For example, `a.reshape((3,-1))` is equivalent to `a.reshape((3,2))`.
    
</p>
</details>

<details><summary><u>Solution.</u></summary>
<p>
    
```python    
a = np.arange(10)

#solution 1
L =[]
for nr in range(1,len(a)+1): #iterate over all possible number of rows
    if len(a)%nr==0: #check if you get an integer number of columns
        L.append(a.reshape((nr,-1))) 
    
#solution 2
L = [a.reshape((nr,-1)) for nr in range(1,len(a)+1) if len(a)%nr==0]
L

```
    
</p>
</details>

### 02 Broadcasting
Create an 3x3 array of ones then:
* multiply each element by 2
* multiply the first column by 1, the second by 2, and the third by 3
* multiply the first row by 1, the second by 2, and the third by 3

Print out the result!

<details><summary><u>Hint.</u></summary>
<p>

According to the rules of broadcasting, two arrays can be multiplied if they have the same shape, or if they can be stretched out to match each other's shape. For example:
* Multiplying arrays with shapes `(4,4)` and `(4,4)` works, because they have the same shape
* Multiplying arrays with shapes `(4,4)` and  by a number works, because the number can be stretched out to match the first array, in effect multiplying each element with the same number.
* Multiplying arrays with shapes `(4,4)` and `(4,)` works, because the second array can be stretched out vertically 
* Multiplying arrays with shapes `(4,4)` and `(1,4)` also works, because the second array can be stretched out horizontally
    
To get an array with shape `(3,)` containing numbers from 1 to 3 use `np.arange(1,4)`. You can convert this to an array with shape `(1,3)` by reshaping it.     
</p>
</details>

<details><summary><u>Solution.</u></summary>
<p>
    
```python
a = np.ones((3,3))
(2*a*np.arange(1,4))*(np.arange(1,4).reshape((3,1)))

```

</details>

### 03 Random segments

What is the expected distance of two randomly selected points in the unit square? Determine it by generating random segments!

You can compare your result to the theoretical value [here](https://mathworld.wolfram.com/SquareLinePicking.html).

<details><summary><u>Hint.</u></summary>
<p>
    
The length of a segment is $l=\sqrt{(x_1-x_2)^2+(y_1-y_2)^2}$.
    
To generate a random point represented by two random coordinates use `np.random.random(size=2)`, to generate an array containing 10000 random points use `np.random.random(size=(n,2))`.

To estimate the expected lenght follow the steps:
* Generate two arrays `coords1` and `coords2` both containing random two dimensional points, these represent the endpoints of a segment.
* Calculate the length of the segment by subracting `coords1` and `coords2`, take the square of the result elementwise, and sum up the rows of the result, finally take the square root elementwise. 
* Repeat this 10000 times and calculate the average.
    
</p>
</details>

<details><summary><u>Solution.</u></summary>
<p>
    
```python
n=10000

#using a for loop
length_sum = 0
for i in range(n):
    coords1 = np.random.random(2)
    coords2 = np.random.random(2)
    length = (((coords1-coords2)**2).sum())**.5
    length_sum += length
avglength = length_sum/n
print("the average length is", avglength)
    
#vectorizing the code makes it much faster
coords1 = np.random.random((n,2))
coords2 = np.random.random((n,2))
lengths = (((coords1-coords2)**2).sum(axis=1))**.5
avglength = lengths.mean()
print("the average length is", avglength)

# or in one line
(((np.random.random((n,2))-np.random.random((n,2)))**2.).sum(axis=1)**.5).mean()

```
    
</p>
</details>

### 05 List of Sundays

Create a list of strings that contain the date of all Sundays of the year 1815 in the format of "June 08, 1815".

<details><summary><u>Hint</u></summary>
<p>

You can loop through all days of a year using a `while` loop by starting from January 1, 1815 and adding one day to the date while the year remains equal to 1815, e.g., using `date.year==1815`.
    
You can test if `date` corresponds to a Sunday using `date.weekday()`.
    
</p>
</details>

<details><summary><u>Solution.</u></summary>
<p>
    
```python
sundays = []
date = datetime.datetime(1815,1,1)
dt = datetime.timedelta(days=1)
while date.year==1815:
    if date.weekday()==6:
        sundays.append(date.strftime("%B %d, %Y"))
    date = date + dt
sundays[:10]
```
    
</p>
</details>
