        return 1
    else:

### 08 Modifying dictionaries

Write two functions that both take a dictionary `D` as input and if `D` contains `1` as a key, changes its value to `one`  but:
* the first one modifies `D` in-place
* the second one leaves `D` unchanged and returns a new dictionary

For example: `{-1:'minus one', 1:'une', 2:'two'}` is changed to `{-1:'minus one', 1:'one', 2:'two'}`

To test if `D` contains `key` as a key use `if key in D`

<details><summary><u>Hint.</u></summary>
<p>

In-place: Dictionaries are mutable, so if you change one of its elements inside the function it will change the original too.
    
New dictionary: to leave the original unchanged, you have to create a new copy of the dictionary inside the function and apply the change to the copy.
   
</p>
</details>

<details><summary><u>Solution.</u></summary>
<p>
    
```python
def mod_inplace(D):
    if 1 in D:
        D[1]='one'
    return
        
#test
D = {-1:'minus one', 1:'une', 2:'two'}
mod_inplace(D)
print(D)

def mod_copy(D):
    D2={}
    for key,value in D.items():
        if key!=1:
            D2[key]=value
        else:
            D2[key]='one'
    
    return D2

#test
D = {-1:'minus one', 1:'une', 2:'two'}
D2 =mod_copy(D)
print(D,D2)
    
#or
def mod_copy2(D):
    D2=D.copy()
    mod_inplace(D2)
    return D2

#test
D = {-1:'minus one', 1:'une', 2:'two'}
D2 =mod_copy2(D)
print(D,D2)
```
    
</p>

D = {}
for key,value in zip(keys,values):

### 12 Dictionary intersection

Write a function that takes two dictionaries `D1` and `D2` as input, and returns a new dictionary that contains the `key:value` pairs that are in both `D1` and `D2`

For example: `D1={'a':1, 'b':42, 'c':-3}` and `D2={'a':0, 'b':42, 'd':-3}` returns `{'b':42}`.

<details><summary><u>Hint.</u></summary>
<p>
    
To test if `D` contains `key` as a key use `if key in D`
    
</p>
</details>

<details><summary><u>Solution.</u></summary>
<p>
    
```python
def dict_intersect(D1,D2):
    D3 = {}
    for key,value in D1.items():
        if key in D2 and D2[key]==value:
            D3[key]=value
    return D3


#test
D1={'a':1, 'b':42, 'c':-3}
D2={'a':0, 'b':42, 'd':-3}
dict_intersect(D1,D2)

```
