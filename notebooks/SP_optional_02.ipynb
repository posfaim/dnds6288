Regular expressions are used to find and manipulate patterns in text. In the following we will learn the basics of regex, and as an example application import a web page (http://www.hydroinfo.hu/tables/ENG/442027H.html) and extract the water level forecast of the Danube.

So how can we use regex to do this?


A regular expression is a string that will be used as a search pattern in another string.

For example, if we have a string: <pre>"This is a test"</pre>, and a regular expression <pre>"is a"</pre>.
This regular expression will 'match' the substring <pre>"is a"</pre> in the first string.

Special characters <pre>. ^ $ * + ? { } [ ] \ | ( )</pre>

The dot character (.) matches any character.

The star character (*) matches any number of repetitions of the pevious character.

The plus character (+) matches any nonzero number repetitions of the previous charcter.

The ? character matches zero or one repetitions of the previous characters.

The ^ character matches the beginning of the string.

The $ character matches the end of the string.

Helpful tool for debugging and understanding regexp: https://regexr.com

````python
import re

text = "This is a test."

pattern = "is a"
res = re.search(pattern, text)
if res is not None:
    print("the match:",res.group())
    print("index of first character of match:", res.start(), "last:",res.end())
else:
    print("No match.")
````

The function `re.search()` returns the first match and its location, and if there is no match, it returns `None`.

Some more interesting examples:

````python
name1 = "MP"
name2 = "János Kertész"
name3 = "P. Howard"

# .= any character
# * = any number of repetitions
res = re.search(r"J.*K", name2)
if res is not None:
    print(res.group(), res.start(), res.end())
else:
    print("No match.")
````

````python
# + = nonzero number of repetitions
res = re.search(r"M.+P", name1) 
if res is not None:
    print(res.group(), res.start(), res.end())
else:
    print("No match.")
````

````python
res = re.search(r"M.*P", name1) 
if res is not None:
    print(res.group(), res.start(), res.end())
else:
    print("No match.")
````

````python
print(r"A\nB")
````

````python
# to match . you have to escape the . in the pattern
res = re.search(r"P.", "Pa Howard") 
if res is not None:
    print(res.group(), res.start(), res.end())
else:
    print("No match.")
````

Note that we used raw strings (e.g., `r"J,*K"`) to specify the patterns, this way we do not have to escape each special character.



The `re.sub(pattern1, pattern2, text)` function matches `pattern1` in `text` and substitutes it with `pattern2`. For example, we can transform first-name-first names to family-name-first:

````python
name2 = "János Kertész"
re.sub(r"(.*) (.*)",r"\2 \1",name2)
````

We can use `()` to specify groups that we can refer to as `\1,\2,...` later, where the number is the order they appear in.

Multiple subtitutions can happen too:

````python
re.sub(r"banana",r"apple","My favorite fruit is the banana, because bananas are great.")
````

Parsing goes from left to right and finds the longest substring that fits the pattern

````python
t = "abracadabra, poof!"

print(re.sub(r"a.*a","",t))
print(re.sub(r"a.{1}a","",t))  # {n} : exactly n repetitions
print(re.sub(r"a.{2}a","",t))  # we can have multiple matches
print(re.sub(r"^a.{2}a","",t)) # ^ = beginning of string
````

`[...]` one of the characters inside the brackets, for example, `[ab]` = either `a` or `b`

You can also give ranges:
* `[0-9]` = any digit
* `[A-Z]` = any capitalized letter

You can also match any character not inside a set using `[^...]`, for example `[^ab]` = anything but `a` or `b`

For example, extracting two-digit numbers:

````python
t = "asda12aasd324asdad"

for res in re.finditer(r"[0-9]{2}", t):
    print(res.group())
````

The `re.finditer()` function allows us to iterate through all matches.

Finding positive or negative integers in a text:

````python
t = "The tallest person in the world was 272 cm tall. Yesterday the temperature was -5 C. A baker's dozen is 13."
for res in re.finditer(r"-?[0-9]+", t): # ? = preceding character is matched once or zero times
    print(res.group())
````

Other useful shorthands:

|Character|Description
|:---|:---
|`\d`| one digit, same as `[0-9]`
|`\w`| word letter = letter, digit, or underscore
|`\s`| white space, such as space, tab, new line

[More info on regex in python.](https://docs.python.org/3.7/library/re.html)

### Extract water level forecasts

Now that we know the basics of regex, we can use it to obtain the timestamps and corresponding water level predicitions from this website:
http://www.hydroinfo.hu/tables/ENG/442027H.html)

First, let's download the contents of the web site:

````python
import urllib.request
link = "http://www.hydroinfo.hu/tables/ENG/442027H.html"
f = urllib.request.urlopen(link) #opens website as a file
content = str(f.read()) # stores entire contents as a string
print(content[:200])
print(len(content))
````

The website is written in `html`, which is used by your browser to render and display what you see. In a later class we will learn more about `html` parsing, now we focus on regular expressions.

`<b>235</b>`.
Write a regular expression that matches
- the date and time in the a table
- and another that matches the water level.

Iterate through the matches and print them out.
